```python
t = tuple(map(int, input().split()))
k = int(input())

n = len(t)
k = k % n  

result = t[k:] + t[:k]

print(result)
```

```python
pairs = input().split()

result = {}

for p in pairs:
    key, value = p.split(":")
    value = int(value)
    
    if key not in result:
        result[key] = []
    
    result[key].append(value)

print(result)
```

W6A5
Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 'zeros': 
...} (đếm số >0, <0, =0)

```python
nums = list(map(int, input().split()))

result = {
    'positives': 0,
    'negatives': 0,
    'zeros': 0
}

for x in nums:
    if x > 0:
        result['positives'] += 1
    elif x < 0:
        result['negatives'] += 1
    else:
        result['zeros'] += 1

print(result)
```

Viết chương trình nhập vào một list các chuỗi, sau đó đếm số lần xuất hiện của từng 
chuỗi và lưu vào dictionary

```python
words = input().split()

result = {}

for w in words:
    if w in result:
        result[w] += 1
    else:
        result[w] = 1

print(result)
```

W6A9
Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới 
trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có 
trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của 
key.

```python
def input_dict():
    pairs = input().split()
    d = {}
    for p in pairs:
        key, value = p.split(":")
        d[key] = int(value)
    return d

d1 = input_dict()
d2 = input_dict()

result = {}
for k in d1:
    result[k] = d1[k]
for k in d2:
    if k in result:
        result[k] += d2[k]
    else:
        result[k] = d2[k]

result_sorted = dict(sorted(result.items()))

print(result_sorted)
```

W6A10
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra 
list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple 
có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của 
số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên 
trong tuple.

```python

lst = list(map(int, input().split()))
k = int(input())

result = []

for i in range(len(lst)):
    for j in range(i + 1, len(lst)):
        if lst[i] + lst[j] == k:
            result.append((lst[i], lst[j]))
def first_element(x):
    return x[0]
result.sort(key=first_element)
print(result)
```
